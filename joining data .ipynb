# Full/Outer Join 

```python
food=pd.read_csv("/Users/madhanmj/Downloads/python 2 data set/SBA-class-activity-main/restaurant_foods.csv")
w1=pd.read_csv("/Users/madhanmj/Downloads/python 2 data set/SBA-class-activity-main/restaurant_week_1_sales.csv")
w2=pd.read_csv("/Users/madhanmj/Downloads/python 2 data set/SBA-class-activity-main/restaurant_week_2_sales.csv")
customers=pd.read_csv("/Users/madhanmj/Downloads/python 2 data set/SBA-class-activity-main/restaurant_customers.csv")
```

```python
w1.head()
```

```python
w1.merge(w2,how="outer",on="Customer ID")
```

```python
w1.merge(w2,how="outer",on="Customer ID",suffixes=[" - Week 1"," - Week 2"])
```

```python
w1.merge(w2,how="outer",on="Customer ID",suffixes=[" - Week 1"," - Week 2"],indicator=True)
```

```python
merged =w1.merge(w2,how="outer",on="Customer ID",suffixes=[" - Week 1"," - Week 2"],indicator=True)
```

```python
merged["_merge"].value_counts()
```

```python
merged[merged["_merge"].isin(["left_only","right_only"])]
```
